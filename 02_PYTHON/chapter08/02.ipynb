```python
class Animal:

    def __init__(self, kind):
        self.kind = kind # 동물의 종류
    def move(self):
        print("동물이 움직인다.")

class Dog(Animal):

    def __init__(self):
        super().__init__("강아지") # 부모 인스턴스 변수인 kind에 초기화
    
    def move(self):
        super().move() #부모쪽 객체의 move() 호출
        print(self.kind + "걸음으로 움직인다.")


class Cat(Animal):
    def __init__(self):
        super().__init__("고양이")

    def move(self):
        print("고양이 걸음으로 움직인다.")
```

```python
dog = Dog()
dog.move()

cat = Cat()
cat.move()
```

```python
print(dog.kind, cat.kind)
```

```python
class Schedule:
    def __init__(self, year, month, day):
        self.__year = year
        self.__month = month
        self.__day = day

    def get_year(self):
        return self.__year

    def get_month(self):
        return self.__month

    def get_day(self):
        return self.__day

    def set_year(self, year):
        self.__year = year

    def set_month(self, month):
        self.__month = month

    def set_day(self, day):
        if self.__month == 2 and day >28: # 값에 대한 통제가 가능
            day = 28
        self.__day = day
    
    def __str__(self):
        return f"year={self.__year}, month={self.__month}, day={self.__day}"
```

```python
s1 = Schedule(2026, 8, 18)

s1.set_month(2)
s1.set_day(31)

print(s1)
```

```python
class SmartLamp:
    def __init__(self, model, is_on=False):
        self.model = model
        self.is_on = is_on

    def turn_on(self):
        self.is_on = True
        return f"{self.model} 조명이 켜졌습니다."

    def turn_off(self):
        self.is_on = False
        return f"{self.model} 조명이 꺼졌습니다."
```

```python
Lamp = SmartLamp("모델명")

print(Lamp.turn_on())
print(Lamp.turn_off())
```

```python
class Hero:
    def __init__(self, name, hp=100):
        self.name = name
        self.hp = hp

    def take_damage(self, damage):
        print(self.hp - damage)
        self.hp -= damage
        if self.hp <= 0:
            print(f"{self.name} 캐릭터가 쓰러졌습니다.")
```

```python
hero = Hero("dd")

hero.take_damage(20)
hero.take_damage(30)
hero.take_damage(50)
```

```python
class DoorLock:

    def __init__(self):
        self.__password = 1232

    def get_password(self):
        return self.__password

    def set_password(self, password):
        if len(password) < 4:
           print("경고: 비밀번호는 4자리 이상이어야 합니다")
           return

        self.__password = password
        print("정상 변경되었습니다")
```

```python
dr = DoorLock()
dr.set_password("1234")
```
