```python
class Bus :
    # 속성 정의
    def __init__(self):
        self.count = 0 # 현재 승객 수, 초기화 : 변수가 되었을 때 최초고 값을 대입
        self.Passengers = []

    def ride(self, Passenger): # 버스 탑승
        self.count +=1 # 승객 수 증가
        self.Passengers.append(Passenger)
        print("탑승")

    def off(self, Passenger): # 하차
        if self.count > 1: # 승객이 마너이스가 될 수 없어서 써 줌
            self.count -= 1  # 승객 수 감소
        self.Passengers.remove(self.Passenger)

    print("하차")
    
    def show_count(self):
        print("승객 수:", self.count)

    def print_self(self):
        print(id(self))

    # def __str__(self):
    #     return f"승객 수: {self.count}, 승객: {self.Passengers}"

    def __repr__(self):
        return "REPR"

class Passenger:
    pass
    # self.name , self.age -> 객체가 생성이 되면 만들어지는 인스턴스 변수
    # (..., name, age) -> 객체를 생성하기 위해서 매개변수로 정의된 지역 변수
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"이름: {self.name}, 나이: {self.age}"

    def __repr__(self):
        return f"Passenger(name='{self.name}', age={self.age})"
```
