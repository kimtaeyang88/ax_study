```python
class Calculator:
    count = 0

    # clz = Calculator 클래스 객체 참조 변수
    # clz가 있는 이유 : 클래스 변수, 다른 클래스 메서드에 접근하기 위한 목적
    @classmethod
    def add(clz, num1, num2):
        clz.count += 1
        print("호출 횟수:", clz.count)
        
        return num1 + num2

    @staticmethod # 참조가 없다.
    def minus(num1, num2):
        return num1 - num2
```

```python
Calculator.add(10, 20)
```

```python
Calculator.minus(10, 20)
```
