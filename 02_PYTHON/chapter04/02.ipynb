```python
total = 0
start = 1

for i in range(100): # 100번 반복
    total += start
    start += 1
    print("반복횟수:", i)
    
print(f"1~100합계: {total}")
```

```python
total = 0
for i in range(1, 101): # 1 ~ 100 사이
    if i % 2 == 0: #짝수
        continue #현재 반복문의 코드 실행을 중단하고 다시 반복 -> 건너뛰기
    total += 1
print(f"합계: {total}")
```
