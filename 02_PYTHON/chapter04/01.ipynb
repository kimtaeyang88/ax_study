```python
age = int(input())
if age >= 20:
    print("성인")
elif age >= 17:
    print("고등학생")
elif age >= 15:
    print("중학생")
elif age >= 8:
    print("초등학생")
else: # 앞선 모든 조건문이 False 이면 실행되는 구문
    print("유치원생")
```
