```python
def generate_nums():

    print("1번째")
    yield 10

    print("2번째")
    yield 20

    print("3번째")
    yield 30
```

```python
gen = generate_nums()
gen
```

```python
next(gen) # 꺼내올 때마다 힙 영역에 해당 값을 할당
```
