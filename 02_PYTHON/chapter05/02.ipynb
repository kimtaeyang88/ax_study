```python
a = True
def outer():
    a='Enclosing'
    print("outer:", a)

    def inner():
        global a # a라는 변수는 전역변수를 찾겠다
        a = not a
        print("inner:", a)
    inner()
outer()
```

```python
def outer(num1):
    def inner(num2):
        return num1 + num2
    
    return inner
```

```python
a = outer(10) # 항상 10을 더해주는 새로운 함수 - num1이  inner 함수에 속박 (closure)
a(20)
```

```python
a(30) # 10+30
```

```python
b = outer(20) # 항상 20을 더해주는 새로운 함수
b(20)
```

```python
def change_numbers(*args, callback):
    numbers = list(args)

    for i in range(len(numbers)):
        numbers[i] = callback(numbers[i])

    return numbers
```

```python
def square(num):
    return num ** 2
    
```

```python
change_numbers(10, 20, 30, callback=square)
```

```python
def plus(num):
    return num + 10

change_numbers(10, 20, 30, callback=plus)
```

```python
square = lambda num: num ** 2
square(10)
```

```python
def change_numbers(callback, *args):
    numbers = list(args)

    for i in range(len(numbers)):
        numbers[i] = callback(numbers[i])

    return numbers
```

```python
def outer():
    num1 = 10

    def inner():
        num2 = 20
        print(num1 + num2)

    return inner
    
inner = outer() # num1은 inner에 의해서 사용되므로 힙 영역에 생성되어 있음
inner() # 10 + 20
```

```python
def add(num1):
    def _add(num2):
        def __add(num3):
            return num1 + num2 + num3

        return __add
    return _add
    
```

```python
result1 = add(10)(20)(30)
result1
```
