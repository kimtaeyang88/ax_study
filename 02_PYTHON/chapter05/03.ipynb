```python
# 재귀 함수 - 같은 함수 내에서 함수를 반복해서 호출하는 방법
def factorial2(num):
    if (num < 1 ):
        return 1
    return num * factorial2(num-1)
```

```python
import time
start = time.time()
result = factorial2(10)
print(result)
end = time.time()
print("걸린 시간:", end - start)
```

```python
import time
def time_checker(callback, num):
    start = time.time() #공통기능

    result = callback(num) #핵심기능 - 팩토리얼 연산 함수(factorial1, factorial2)

    end = time.time() #공통기능
    print("걸린시간:", end-start)
    return result
```

```python
time_checker(factorial2, 10)
```

```python
import time
def time_checker2(callback):
    def wrapper(num1, num2):
        start = time.time()
        result = callback(num1, num2) # 핵심 기능
        end = time.time()
        print("걸린시간:", end-start)

        return result
    
    return wrapper
```

```python
@time_checker2
def factorial2(num):
    if (num < 1 ):
        return 1
    return num * factorial2(num-1)
```

```python
factorial2(10)
```

```python
@time_checker2
def add(num1, num2):
    return num1 + num2
```

```python
add(10,20)
```
