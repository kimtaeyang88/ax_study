```python
def login(user_id, user_pass):
    if user_id != "user01":
        raise ValueError("아이디가 일치하지 않습니다.") # 예외가 발생 - 코드 실행 흐름은 중단

    if user_pass != "1234":
        raise ValueError("비밀번호가 일치하지 않습니다")

    print("로그인 처리 로직...")
```

```python
try:
    login("user01", "1234")
except ValueError as e:
    print(e)
    print("회원 페이지에서 추가적으로 할 로직...")
```

```python
class UserIDMissmatchError(Exception):
    def __str__(self):
        return "아이디가 일치하지 않습니다."

class UserPassMissmatchError(Exception):
    def __str__(self):
        return "비밀번호가 일치하지 않습니다."
```

```python
def login(user_id, user_pass):
    if user_id != "user01":
        raise UserIDMissmatchError() # 예외가 발생 - 코드 실행 흐름은 중단

    if user_pass != "1234":
        raise UserPassMissmatchError()

    print("로그인 처리 로직...")
```

```python
order = input()

try:

    foodlist = ("피자", "치킨", "햄버거") 
   
    result = int(order)  # result 는 입력받은 값을 숫자로 바꾸는 것이고 숫자는 foodlist에서 내용물을 가지고 오기 위한 것 = index
    
    # foodlist 는 배열. index를 사용해서 foodlist 안의 값을 가져오기 위한 식
    # foodlist[result]
    #사용자가 1을 입력하면 치킨을 가져온다
    # foodlist[1]

    menu = foodlist[result]

except ValueError:
    print("숫자만 입력하세요")

except IndexError:
    print("없는 번호입니다")

else:
    print(f"{menu} 주문하셨습니다.")

finally:
    print("종료")
```

```python
def calculate_share(total_price, num_people):
    share_per_person = total_price / num_people
    return share_per_person

def start_settlement():
    participants = 0
    result = calculate_share(50000, participants)
    print(result)

start_settlement()
```

```python
def calculate_revenue():
    total_revenue = 0
    daily_sales = [12000, 25000, 31000]
    for amount in daily_sales:
        total_revenue += amount
    print(total_revenue)

calculate_revenue()
```

```python
def sum_scores():
    final_score = 50
    bonus_scores = [10, 20, 30]
    for score in bonus_scores:
        final_score += score
    return final_score

result = sum_scores()
```
